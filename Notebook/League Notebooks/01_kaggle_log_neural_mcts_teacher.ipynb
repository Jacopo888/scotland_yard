```python
%%bash
set -e
cd /kaggle/working
python -m pip install -q networkx numpy pandas pyarrow tqdm scipy matplotlib torch || true
REPO_DIR=/kaggle/working/scotland_yard
if [ ! -d "$REPO_DIR" ]; then
  git clone --depth 1 https://github.com/Jacopo888/scotland_yard.git "$REPO_DIR"
fi
cd "$REPO_DIR"
python model_registry.py --verify-paths
```

```python
GAMES = 2000
SIMULATIONS = 32
DEVICE = 'cuda'
SEED_BASE = 20260522
OUTPUT_DIR = '/kaggle/working/neural_mcts_logs'

print('games', GAMES, 'simulations', SIMULATIONS, 'device', DEVICE)
```

```python
import subprocess, sys

cmd = [
    sys.executable, 'league/neural_mcts_logger.py',
    '--games', str(GAMES),
    '--simulations', str(SIMULATIONS),
    '--device', DEVICE,
    '--seed-base', str(SEED_BASE),
    '--output-dir', OUTPUT_DIR,
    '--games-per-part', '100',
    '--log-every', '25',
]
subprocess.run(cmd, cwd='/kaggle/working/scotland_yard', check=True)
```
