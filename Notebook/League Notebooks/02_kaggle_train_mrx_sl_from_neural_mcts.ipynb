```python
DATA_DIR = None  # None = prende l'ultimo /kaggle/working/neural_mcts_logs/neural_mcts_logs_*
EPOCHS = 12
BATCH_SIZE = 256
DEVICE = 'cuda'
OUTPUT_DIR = '/kaggle/working/mrx_sl_checkpoints'

print('data_dir', DATA_DIR or 'auto-latest')
```

```python
import subprocess, sys

cmd = [
    sys.executable, 'league/train_mrx_sl.py',
    '--epochs', str(EPOCHS),
    '--batch-size', str(BATCH_SIZE),
    '--device', DEVICE,
    '--output-dir', OUTPUT_DIR,
]
if DATA_DIR:
    cmd += ['--data-dir', DATA_DIR]
subprocess.run(cmd, cwd='/kaggle/working/scotland_yard', check=True)
```
